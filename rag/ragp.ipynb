```python
import fitz

def extract_pdf(pdf_path):
    doc = fitz.open(pdf_path)

    documents = []

    for page_num in range(doc.page_count):
        page = doc[page_num]

        text = page.get_text()

        if text.strip():
            documents.append({
                "page": page_num + 1,
                "text": text
            })

    return documents
```

```python
docs = extract_pdf("data/MSFT2024.pdf")

print("Pages extracted:", len(docs))
print(docs[0]["text"][:2000])
```
